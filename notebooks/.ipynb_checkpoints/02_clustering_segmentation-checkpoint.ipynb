# Customer Segmentation via RFM + K-Means Clustering
## Enterprise Predictive Analytics Engine — E-commerce Domain

**Objective:** Segment customers into meaningful groups based on their
purchasing behavior, using RFM (Recency, Frequency, Monetary) features
and K-Means clustering.

**Input:** `data/processed/olist_master_cleaned.csv`
**Output:** `data/processed/customer_segments.csv`

**Steps:**
1. Load cleaned data
2. Build RFM features per customer
3. Scale features
4. Find optimal number of clusters (elbow method)
5. Run K-Means and label segments
6. Visualize and interpret segments
7. Save segmented output

## 1. Load Cleaned Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

sns.set_style("whitegrid")

df = pd.read_csv("../data/processed/olist_master_cleaned.csv", parse_dates=["order_purchase_timestamp"])
print("Shape:", df.shape)
df.head()
```

## 2. Build RFM Features

- **Recency:** days since customer's last purchase (relative to most recent date in dataset)
- **Frequency:** number of distinct orders placed
- **Monetary:** total amount spent

We use `customer_unique_id` rather than `customer_id`, since Olist assigns
a new `customer_id` per order — `customer_unique_id` is the true customer identifier.

```python
snapshot_date = df["order_purchase_timestamp"].max() + pd.Timedelta(days=1)

rfm = df.groupby("customer_unique_id").agg(
    recency=("order_purchase_timestamp", lambda x: (snapshot_date - x.max()).days),
    frequency=("order_id", "nunique"),
    monetary=("payment_value", "sum")
).reset_index()

print("RFM table shape:", rfm.shape)
rfm.describe()
```

## 3. Scale Features

K-Means is distance-based, so features must be on the same scale.
We standardize Recency, Frequency, and Monetary before clustering.

```python
rfm_features = rfm[["recency", "frequency", "monetary"]]

scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_features)

rfm_scaled = pd.DataFrame(rfm_scaled, columns=["recency", "frequency", "monetary"])
rfm_scaled.head()
```

## 4. Find Optimal Number of Clusters (Elbow Method)

We test cluster counts from 2 to 10 and plot inertia (within-cluster
sum of squares) to identify the "elbow" point — where adding more
clusters stops meaningfully improving the fit.

```python
import os
os.makedirs("../outputs/figures", exist_ok=True)

inertia = []
K_range = range(2, 11)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(rfm_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K_range, inertia, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal k")
plt.savefig("../outputs/figures/elbow_method.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 5. Run K-Means and Label Segments

Based on the elbow plot, we choose k=4 — a practical number of segments
for business interpretation (too many clusters become hard to act on).

```python
optimal_k = 4

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
rfm["cluster"] = kmeans.fit_predict(rfm_scaled)

# Inspect average RFM values per cluster to interpret them
cluster_summary = rfm.groupby("cluster")[["recency", "frequency", "monetary"]].mean().round(1)
cluster_summary["count"] = rfm["cluster"].value_counts()
cluster_summary
```

## 6. Interpret and Name Segments

Based on the cluster summary, we assign human-readable business labels:
- **Cluster 0** (recent, one-time, avg spend) → "Recent One-Time Buyers"
- **Cluster 1** (long inactive, one-time, avg spend) → "Lapsed / At Risk"
- **Cluster 2** (repeat buyers, higher spend) → "Loyal Repeat Customers"
- **Cluster 3** (tiny group, extreme spend) → "High-Value Outliers"

```python
segment_map = {
    0: "Recent One-Time Buyers",
    1: "Lapsed / At Risk",
    2: "Loyal Repeat Customers",
    3: "High-Value Outliers"
}

rfm["segment"] = rfm["cluster"].map(segment_map)
rfm["segment"].value_counts()
```

## 7. Visualize Segments

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))

sns.boxplot(data=rfm, x="segment", y="recency", ax=axes[0])
axes[0].set_title("Recency by Segment")
axes[0].tick_params(axis='x', rotation=30)

sns.boxplot(data=rfm, x="segment", y="frequency", ax=axes[1])
axes[1].set_title("Frequency by Segment")
axes[1].tick_params(axis='x', rotation=30)

sns.boxplot(data=rfm, x="segment", y="monetary", ax=axes[2])
axes[2].set_title("Monetary by Segment")
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.savefig("../outputs/figures/segment_boxplots.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 8. Save Segmented Customer Data

This output will feed into the classification notebook (predicting
churn/risk per segment) and the Power BI dashboard.

```python
rfm.to_csv("../data/processed/customer_segments.csv", index=False)
print("Saved to data/processed/customer_segments.csv")
rfm.head()
```
